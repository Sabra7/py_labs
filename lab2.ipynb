3.6 if..else and if..elif..else Statements

```python
grade = int(input("Enter your grade"))

if grade >= 90 :
   print("Your grade above 'A'")
elif grade >= 80 :
   print("Your grade above 'B'")
elif grade >= 70 :
   print("Your grade above 'C'")
elif grade >= 60 :
   print("Your grade above'D'")
else : print ("Your failed")
```

```python
age = int(input("Enter your age "))
if age >= 18 and age <= 60:
    print("The person's age is between 18 and 60.")
else:
    print("The person's age is NOT between 18 and 60.")
```
