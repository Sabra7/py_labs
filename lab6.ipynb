```python
import numpy as np
```

7.2 Creating Arrays from Existing Data

```python
my_list = [1, 2, 3, 4, 5]
arr = np.array(my_list)
print(arr)
```

7.3 Array Attributes

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])

print("ndim:", arr.ndim)
print("shape:", arr.shape)
print("size:", arr.size)
print("dtype:", arr.dtype)
```

7.4 Filling Arrays with Specific Values

```python
zeros_array = np.zeros((3, 4))
print(zeros_array)
```

7.5 Creating Arrays from Ranges

```python
arr = np.arange(10, 101, 10)
print(arr)
```

7.6 List vs. Array Performance: Introducing %timeit

```python
list1 = list(range(100000))
list2 = list(range(100000))
%timeit [x + y for x, y in zip(list1, list2)]

arr1 = np.arange(100000)
arr2 = np.arange(100000)
%timeit arr1 + arr2
```

7.7 Array Operators

```python
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([10, 20, 30, 40])

print("Addition:", arr1 + arr2)
print("Multiplication:", arr1 * arr2)
```

7.8 NumPy Calculation Methods

```python
arr = np.array([10, 25, 5, 40, 30])

print("Minimum:", arr.min())
print("Maximum:", arr.max())
print("Mean:", arr.mean())
print("Sum:", arr.sum())
```

7.9 Universal Functions

```python
arr = np.array([1, 4, 9, 16, 25])
result = np.sqrt(arr)

print(result)
```

7.10 Indexing and Slicing

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("First row:", arr[0, :])
print("Last column:", arr[:, -1])
```

7.11 Views: Shallow Copies

```python
original = np.array([10, 20, 30, 40])
view_arr = original.view()

view_arr[0] = 999

print("View:", view_arr)
print("Original:", original)
```

7.12 Deep Copies

```python
original = np.array([10, 20, 30, 40])
copy_arr = original.copy()

copy_arr[0] = 999

print("Copy:", copy_arr)
print("Original:", original)
```

7.13 Reshaping and Transposing

```python
arr = np.arange(12)
reshaped = arr.reshape(3, 4)

print("Reshaped (3x4):\n", reshaped)
print("\nTransposed:\n", reshaped.T)
```
