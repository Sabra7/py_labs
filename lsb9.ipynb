Q1. Write a program that plots the values [1, 4, 9, 16, 25] using a line chart.

```python
import matplotlib.pyplot as plt

plt.plot([1, 4, 9, 16, 25])
plt.show()
```

Q2. Write a program that plots two lines on the same graph with appropriate labels.

```python
y1 = [1, 2, 3, 4, 5]
y2 = [1, 4, 9, 16, 25]

plt.plot(y1, label="Line 1")
plt.plot(y2, label="Line 2")
plt.show()
```

Q3. Write a program that plots a line using circle markers and a dashed line style.

```python
plt.plot([1, 4, 9, 16, 25], marker="o", linestyle="--")
plt.show()
```

Q4. Write a program that creates a line chart with a title and labels for both axes.

```python
plt.plot([1, 4, 9, 16, 25])
plt.title("Sample Line Chart")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

Q5. Write a program that displays a legend for two plotted lines.

```python
plt.plot([1, 2, 3, 4], label="Dataset A")
plt.plot([4, 3, 2, 1], label="Dataset B")
plt.legend()
plt.show()
```

Q6. Write a program that displays a grid on a line chart.

```python
plt.plot([1, 4, 9, 16, 25])
plt.grid(True)
plt.show()
```

Q7. Write a program that creates a bar chart showing the sales of five products.

```python
products = ["A", "B", "C", "D", "E"]
sales = [100, 150, 80, 200, 120]

plt.bar(products, sales)
plt.show()
```

Q8. Write a program that creates a pie chart showing the percentage of students in three departments.

```python
departments = ["CS", "IS", "IT"]
students = [150, 100, 50]

plt.pie(students, labels=departments, autopct="%1.1f%%")
plt.show()
```

Q9. Write a program that creates a histogram for a list of exam scores.

```python
scores = [55, 60, 65, 70, 75, 80, 85, 90, 95, 100, 85, 90, 75]

plt.hist(scores, bins=5)
plt.show()
```

Q10. Write a program that creates a scatter plot showing the relationship between students' study hours and exam scores.

```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
scores = [50, 55, 65, 70, 75, 85, 90, 95]

plt.scatter(hours, scores)
plt.show()
```

Q11. Write a program that creates two subplots: one line chart and one bar chart.

```python
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3], [4, 5, 6])

plt.subplot(1, 2, 2)
plt.bar(["A", "B"], [10, 20])

plt.show()
```

Q12. Write a program that creates a line chart and saves it as plot.png.

```python
plt.plot([1, 4, 9, 16, 25])
plt.savefig("plot.png")
plt.show()
```
