##file
Q1. Write a program that writes five names to a text file.

```python
names = ["Ali", "Sara", "Khaled", "Mona", "Omar"]
with open("names.txt", "w") as f:
    for name in names:
        f.write(name + "\n")
```

Q2. Write a program that reads and displays the contents of a text file.

```python
with open("names.txt", "r") as f:
    content = f.read()
    print(content)
```

Q3. Write a program that appends a new line to a text file.

```python
with open("names.txt", "a") as f:
    f.write("Fahad\n")
```

Q4. Write a program that counts the number of lines in a text file.

```python
with open("names.txt", "r") as f:
    lines = f.readlines()
    print("Total lines:", len(lines))
```

Q5. Write a program that copies the contents of one text file to another using the with statement.

```python
with open("names.txt", "r") as src, open("copy.txt", "w") as dest:
    dest.write(src.read())
```

# Exceptoin
Q1. Write a program that handles a ZeroDivisionError when dividing two numbers.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

Q2. Write a program that handles a ValueError when converting user input to an integer.

```python
try:
    num = int(input("Enter an integer: "))
    print("Number:", num)
except ValueError:
    print("Error: Invalid integer input.")
```

Q3. Write a program that uses try, except, and finally to read an integer from the user.

```python
try:
    num = int(input("Enter an integer: "))
    print("Entered:", num)
except ValueError:
    print("Error: That is not a valid integer.")
finally:
    print("Execution completed.")
```

Q4. Write a program that handles a FileNotFoundError when opening a text file.

```python
try:
    with open("non_existent_file.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("Error: File not found.")
```

Q5. Write a program that uses try, except, else, and finally to divide two numbers safely.

```python
try:
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))
    result = a / b
except ZeroDivisionError:
    print("Error: Division by zero.")
except ValueError:
    print("Error: Invalid numeric input.")
else:
    print("Result:", result)
finally:
    print("Program finished.")
```
