##7.14.1 pandas Series

Write a program that creates a Series from a list of five integers.

```python
s = pd.Series([10, 20, 30, 40, 50])
print(s)
```

Write a program that creates a Series with custom index labels and displays the value of a given label.

```python
s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(s['b'])
```

Write a program that creates a Series from a dictionary.

```python
data = {'a': 100, 'b': 200, 'c': 300}
s = pd.Series(data)
print(s)
```

Write a program that filters a Series to display only values greater than 50.

```python
s = pd.Series([10, 60, 30, 80, 45, 90])
print(s[s > 50])
```

Write a program that calculates the minimum, maximum, and average values of a Series.

```python
s = pd.Series([10, 20, 30, 40, 50])

print("Minimum:", s.min())
print("Maximum:", s.max())
print("Average:", s.mean())
```

##7.14.2 DataFrames

Write a program that creates a DataFrame from a dictionary of student names and grades.

```python
data = {
    'Name': ['Ali', 'Sara', 'Khaled', 'Mona'],
    'Grade': [85, 92, 78, 95]
}
df = pd.DataFrame(data)
print(df)
```

Write a program that uses loc to display the grade of a specific student.

```python
data = {'Grade': [85, 92, 78]}
df = pd.DataFrame(data, index=['Ali', 'Sara', 'Khaled'])

print(df.loc['Sara', 'Grade'])
```

Write a program that uses iloc to display the first two rows of a DataFrame.

```python
data = {'Name': ['Ali', 'Sara', 'Khaled'], 'Grade': [85, 92, 78]}
df = pd.DataFrame(data)

print(df.iloc[:2])
```

Write a program that uses at to update the grade of a student.

```python
data = {'Grade': [85, 92, 78]}
df = pd.DataFrame(data, index=['Ali', 'Sara', 'Khaled'])

df.at['Sara', 'Grade'] = 98
print(df)
```

Write a program that uses iat to modify a value by its row and column positions.

```python
data = {'Name': ['Ali', 'Sara'], 'Grade': [85, 92]}
df = pd.DataFrame(data)

df.iat[0, 1] = 90
print(df)
```

Write a program that adds a new column named Total to a DataFrame.

```python
data = {'Name': ['Ali', 'Sara'], 'Grade': [85, 92]}
df = pd.DataFrame(data)

df['Total'] = [100, 100]
print(df)
```

Write a program that sorts a DataFrame by the Grade column.

```python
data = {'Name': ['Ali', 'Sara', 'Khaled'], 'Grade': [85, 92, 78]}
df = pd.DataFrame(data)

df_sorted = df.sort_values(by='Grade')
print(df_sorted)
```

Write a program that filters students whose grade is greater than or equal to 90.

```python
data = {'Name': ['Ali', 'Sara', 'Khaled', 'Mona'], 'Grade': [85, 92, 78, 90]}
df = pd.DataFrame(data)

filtered_df = df[df['Grade'] >= 90]
print(filtered_df)
```
