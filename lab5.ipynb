6.2.7 Example: Word Counts: Write a program that counts the frequency of each word in a sentence using a dictionary.


```python
text = ('this is sample text with several words ' 'this is more sample text with some different words')
```

```python
word_counts = {}

for word in text.split():
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print(f'{"WORD":<12}COUNT')

for word, count in sorted(word_counts.items()):
    print(f'{word:<12}{count}')

print('\nNumber of unique words:', len(word_counts))
```
