# Create the dataset and Import the tools

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'country': ['Saudi Arabia', 'Saudi Arabia', 'Saudi Arabia', 'Saudi Arabia',
                'United Arab Emirates', 'United Arab Emirates', 'United Arab Emirates', 'United Arab Emirates',
                'Egypt', 'Egypt', 'Egypt', 'Egypt',
                'United States', 'United States', 'United States', 'United States',
                'Japan', 'Japan', 'Japan', 'Japan'],
    'year': [2019, 2020, 2021, 2022,
             2019, 2020, 2021, 2022,
             2019, 2020, 2021, 2022,
             2019, 2020, 2021, 2022,
             2019, 2020, 2021, 2022],
    'population': [34268529, 35018116, 35950396, 36408820,
                   9599353, 9749660, 9918200, 10020000,
                   100388073, 102118600, 104258300, 106000000,
                   328239523, 331501080, 331893745, 333287557,
                   126264930, 125836021, 125507472, 125124000],
    'gdp': [793, 700, 833, 1108,
            417, 358, 415, 507,
            303, 383, 404, 476,
            21372, 20893, 23315, 25462,
            5117, 5048, 5034, 4231]
}
pd.DataFrame(data).to_csv("world_gdp_population.csv", index=False)

df = pd.read_csv("world_gdp_population.csv")
```

#Display the first 5 rows

```python
print("First 5 Rows ")
display(df.head())
```


# Show the number of rows and columns


```python
print("Dataset Shape (Rows, Columns): ")
print(f"Shape: {df.shape}")
```


#Display column names


```python
print(" Column Names: ")
print(df.columns.tolist())
```

# Clean the data


```python
df.columns = df.columns.str.strip().str.lower()
df = df.dropna()
print(f"Dataset Shape After Cleaning: {df.shape}")
```


#List all unique countries in the dataset


```python
print("Unique Countries ")
print(df['country'].unique())
```

# Average population per country

```python

print(" Average Population per Country ")
avg_pop = df.groupby('country')['population'].mean()
print(avg_pop)
```


# Average GDP per country


```python
print("\n Average GDP per Country ")
avg_gdp = df.groupby('country')['gdp'].mean()
print(avg_gdp)
```


#Summary statistics


```python
print("\n Summary Statistics ")
display(df.describe())
```

#Line Chart: Population Trend for Saudi Arabia (2019–2022)

```python
saudi_df = df[df['country'] == 'Saudi Arabia']

plt.figure(figsize=(8, 4))
plt.plot(saudi_df['year'], saudi_df['population'], marker='o', color='green', linewidth=2)
plt.title("Population Trend for Saudi Arabia (2019-2022)", fontsize=14)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Population", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

#Bar Chart: Average GDP per country

```python
plt.figure(figsize=(8, 4))
avg_gdp.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title("Average GDP per Country", fontsize=14)
plt.xlabel("Country", fontsize=12)
plt.ylabel("Average GDP", fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
