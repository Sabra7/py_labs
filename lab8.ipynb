8.2.1 Presentation Types

```python
n = 10
f = 3.14159
s = "Hello"
print(f"{n:d}, {f:.2f}, {s:s}")
```

8.2.2 Field Widths and Alignment

```python
students = [("Ali", 95), ("Sara", 88), ("Khaled", 92)]
for name, grade in students:
    print(f"{name:<10}{grade:>5}")
```

8.2.3 Numeric Formatting

```python
num = 12.34567
print(f"{num:.2f}")
```

8.2.4 String's format() Method

```python
name = "Ali"
gpa = 3.8
print("Name: {}, GPA: {}".format(name, gpa))
```

8.3 Concatenating and Repeating Strings

```python
first_name = "Ali"
last_name = "Ahmed"
print(first_name + " " + last_name)
print("*" * 30)
```

8.4 Stripping Whitespace from Strings

```python
text = input("Enter a string: ").strip()
print(text)
```

8.5 Changing Character Case

```python
sentence = input("Enter a sentence: ")
print("Upper:", sentence.upper())
print("Lower:", sentence.lower())
print("Title:", sentence.title())
```

8.6 Comparison Operators for Strings

```python
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

if s1 == s2:
    print("Strings are equal")
elif s1 < s2:
    print(f"'{s1}' comes first")
else:
    print(f"'{s2}' comes first")
```

8.7 Searching for Substrings

```python
sentence = "Python is a great language"
word = "Python"
print(word in sentence)
```

8.8 Replacing Substrings

```python
sentence = "Python is fun and Python is powerful"
print(sentence.replace("Python", "Java"))
```

8.9 Splitting and Joining Strings

```python
sentence = "Python is fun to learn"
words = sentence.split()
print("-".join(words))
```

8.10 Characters and Character-Testing Methods

```python
text = input("Enter text: ")
print("Digits only:", text.isdigit())
print("Letters only:", text.isalpha())
print("Alphanumeric:", text.isalnum())
```
